```python
## Import the libraries
```

```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
from pathlib import Path 
```

```python
## Load the Excel file
```

```python
file_path = r"C:\Online_Retail_Data_Analysis\data\Online_Retail.xlsx"

df = pd.read_excel(file_path)

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
## Understand the dataset
```

```python
print("head")
display(df.head())
print("tail")
display(df.tail())

print("rows:" , df.shape)
print("columns:", df.shape)

print(df.columns.tolist())

df.head()
```

```python
display(df.describe())
display(df.describe(include = "all"))
```

```python
## Check missing values
```

```python
missing_values = df.isnull().sum()

missing_summary = pd.DataFrame({
    "missing_count" : missing_values,
    "missing_percentage":(
        missing_values / len(df) * 100).round(2)
})
display(missing_summary)
```

```python
## Check duplicate rows
```

```python
print("duplicated rows:" ,df.duplicated().sum())
```

```python
df = df.drop_duplicates().copy()

print("shape after removing duplicates:", df.shape)
```

```python
df.columns
```

```python
## Handle missing descriptions
```

```python
df["Description"] = (
    df["Description"]
    .fillna("unknown product")
    .astype(str)
    .str.strip()
)
print("missing description:",df["Description"].isnull().sum())
```

```python
## Handle customer IDs
```

```python
print("missing customer IDs:", df["CustomerID"].isnull().sum())

df["CustomerID"] = df["CustomerID"].astype("Int64")

print(df["CustomerID"].dtype)
```

```python
##  invalid quantities and prices
```

```python
print("Quantity <= 0 :" , (df["Quantity"] <= 0).sum())
print("UnitPrice <0:" ,(df["UnitPrice"] <- 0).sum())

print("cancellation invoices:",
      df["InvoiceNo"].astype(str).str.startswith("C").sum())
```

```python
df["InvoiceNo"] = df["InvoiceNo"].astype("string").str.strip()
df["StockCode"] = df["StockCode"].astype("string").str.strip()
df["Country"] = df["Country"].astype("string").str.strip()

valid_invoice = ~df["InvoiceNo"].str.startswith("C", na=False)

valid_quantity = df["Quantity"] > 0
valid_price = df["UnitPrice"] > 0

df = df.loc[
    valid_invoice & valid_quantity & valid_price
].copy()

print("New dataset shape:", df.shape)

display(df.head())
```

```python
## Check data types
```

```python
# Convert InvoiceDate to datetime
df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"],
    errors="coerce"
)

# Convert Quantity and UnitPrice to numeric
df["Quantity"] = pd.to_numeric(
    df["Quantity"],
    errors="coerce"
)

df["UnitPrice"] = pd.to_numeric(
    df["UnitPrice"],
    errors="coerce"
)

# Remove rows with missing essential transaction values
df = df.dropna(
    subset=[
        "InvoiceNo",
        "StockCode",
        "InvoiceDate",
        "Quantity",
        "UnitPrice",
        "Country"
    ]
).copy()

# Convert CustomerID to nullable integer
df["CustomerID"] = df["CustomerID"].astype("Int64")

# Display dataset information
df.info()
```

```python
## Create a revenue column
```

```python
df["Revenue"] = df["Quantity"] * df["UnitPrice"]

df["Revenue"] = df["Revenue"].round(2)

display(df.head)
print("total line item revenue:" ,round(df["Revenue"].sum(),2))
```

```python
## date-based features
```

```python
df["Year"] = df["InvoiceDate"].dt.year
df["Month"] = df["InvoiceDate"].dt.month
df["MonthName"] = df["InvoiceDate"].dt.strftime("%B")
df["YearMonth"] = df["InvoiceDate"].dt.to_period("M").astype(str)
df["Day"] = df["InvoiceDate"].dt.day
df["DayofWeek"] = df["InvoiceDate"].dt.day_name()
df["Hour"] = df["InvoiceDate"].dt.hour

display(
    df[[
        "InvoiceDate","Year","Month","MonthName",
        "YearMonth","Day","DayofWeek","Hour"
    ]].head()
)
```

```python
print("final shape:", df.shape)
print("\n missing values:")
display(df.isnull().sum())

print("\n exact duplicate rows:", df.duplicated().sum())

print("\n non-positive quantities:", (df["Quantity"] <= 0).sum())
print("\n non-positive prices:", (df["UnitPrice"]<= 0).sum())

print("\n date range:")
print(df["InvoiceDate"].min(), "to", df["InvoiceDate"].max())

print("\nTotal revenue:", round(df["Revenue"].sum(), 2))
```

# SQL TO PYTHON CONNECTION 

```python
from sqlalchemy import create_engine, text

DB_USER = "root"
DB_PASSWORD = "root"
DB_HOST = "localhost"
DB_PORT = 3306
DB_NAME = "online_retail_db"

connection_url = (
    f"mysql+pymysql://{DB_USER}:{DB_PASSWORD}"
    f"@{DB_HOST}:{DB_PORT}/{DB_NAME}"
)

engine = create_engine(connection_url)

with engine.connect() as connection:
    result = connection.execute(text("SELECT DATABASE()"))
    print("Connected to database:", result.scalar())
```

```python
sql_columns = [
    "InvoiceNo",
    "StockCode",
    "Description",
    "Quantity",
    "InvoiceDate",
    "UnitPrice",
    "CustomerID",
    "Country",
    "Revenue",
    "Year",
    "Month",
    "MonthName",
    "YearMonth",
    "Day",
    "DayofWeek",
    "Hour"
]

df_sql = df[sql_columns].copy()

df_sql["CustomerID"] = df_sql["CustomerID"].astype(object)
df_sql = df_sql.astype(object).where(
    pd.notna(df_sql), None
)

df_sql.to_sql(
    name="retail_transactions",
    con=engine,
    if_exists="append",
    index=False,
    chunksize=5000,
    method="multi"
)

print("Data loaded into MySQL successfully!")
```

```python
from sqlalchemy import text

with engine.connect() as connection:
    row_count = connection.execute(
        text("SELECT COUNT(*) FROM retail_transactions")
    ).scalar()

print("Rows in SQL:", row_count)
print("Rows in Python:", len(df))
```

# Python exploratory data analysis (EDA)

```python
## Calculate key business metrics
```

```python
total_revenue = df["Revenue"].sum()
total_invoices = df["InvoiceNo"].nunique()
total_customer = df["CustomerID"].nunique()
total_products = df["StockCode"].nunique()
total_countries = df["Country"].nunique()

average_invoice_value = (
    df.groupby("InvoiceNo")["Revenue"].sum().mean()
)

print("total revenue:" , round(total_revenue,2))
print("total invoices:" , total_invoices)
print("total customers:" , total_customer)
print("total products:" , total_products)
print("total countries:", total_countries)
print("average invoice values:", round(average_invoice_value, 2))

    
```

```python
## Analyze monthly revenue
```

```python
monthly_sales = (
    df.groupby("YearMonth", as_index = False)["Revenue"].sum().sort_values("YearMonth")
)
display(monthly_sales)
```

```python
## Analyze monthly revenue
```

```python
monthly_sales = (
    df.groupby("YearMonth",as_index = False)["Revenue"]
    .sum()
    .sort_values("YearMonth")
)
display(monthly_sales)
    
```

```python
## month with the highest revenue:
```

```python
best_month = monthly_sales.loc[
    monthly_sales["Revenue"].idxmax()
    ]
print("best month:",best_month["YearMonth"])
print("revenue:",round(best_month["Revenue"],2))
```

```python
## Analyze country-wise sales
```

```python
country_sales = (
    df.groupby("Country",as_index = False).agg(
        total_revenue = ("Revenue","sum"),
        total_quantity = ("Quantity","sum"),
        total_invoices = ("InvoiceNo","nunique")
        )
        .sort_values("total_revenue",ascending = False)
    )
display(country_sales.head(10))
                       
```

```python
## the highest-revenue country:
```

```python
top_country = country_sales.iloc[0]
print("top country:",top_country["Country"])
print("Revenue:",round(top_country["total_revenue"],2))
```

```python
## Analyze product performance
```

```python
product_sales = (
    df.groupby("StockCode",as_index = False)
    .agg(
        ProductName = ("Description" ,"first"),
        Total_Quantity = ("Quantity","sum"),
        Total_Revenue = ("Revenue","sum")
        )
        .sort_values("Total_Revenue", ascending = False)
)
display(product_sales.head(10))
```

```python
top_products_by_quantity = (
    product_sales.sort_values(
        "Total_Quantity",ascending = False
    ).iloc[0]
)
print("product:", top_products_by_quantity["ProductName"])
print("units sold:", top_products_by_quantity["Total_Quantity"])
```

```python
## Analyze customer behavior
```

```python
customer_df = df.dropna(subset= ["CustomerID"]).copy()

customer_sales = (
    customer_df.groupby("CustomerID", as_index = False)
    .agg
    (
        Total_Revenue = ("Revenue" ,"sum"),
        Total_Orders = ("InvoiceNo","nunique"),
        Total_Quantity = ("Quantity","sum")
    )
    .sort_values("Total_Revenue", ascending = False)
)
display(customer_sales.head(10))
```

```python
repeat_customers = customer_sales[
    customer_sales["Total_Orders"]>1
    ]

repeat_customer_count = len(repeat_customers)
identified_customer_count = len(customer_sales)

repeat_customer_percentage = (
    repeat_customer_count/identified_customer_count*100
    if identified_customer_count else 0
)

print("indentied customers:",identified_customer_count)
print("repeat customers:", repeat_customer_count)
print("repeat customer percentage:",round(repeat_customer_percentage,2),"%")
```

```python
### Sales by day of the week
```

```python
weekday_order = [
    "Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"
]

weekday_sales = (
    df.groupby("DayofWeek", as_index = False)["Revenue"]
    .sum()
)
weekday_sales["DayofWeek"] = pd.Categorical(
    weekday_sales["DayofWeek"],
    categories = weekday_order,
    ordered = True
)
weekday_sales = weekday_sales.sort_values("DayofWeek")

fig, ax = plt.subplots(figsize=(10, 5))

sns.barplot(
    data=weekday_sales,
    x="DayofWeek",
    y="Revenue",
    ax=ax
)

ax.set_title("Revenue by Day of the Week")
ax.set_xlabel("Day")
ax.set_ylabel("Revenue")
ax.tick_params(axis="x", labelrotation=45)

plt.tight_layout()

plt.savefig(
    output_dir/"weekday_sales.png",
    dpi = 300,
    bbox_inches = "tight"
)
plt.show()
```

```python
### Sales by hour
```

```python
hourly_sales = (
    df.groupby("Hour", as_index = False)["Revenue"]
    .sum()
)
fig, ax = plt.subplots(figsize=(12, 5))

sns.barplot(
    data=hourly_sales,
    x="Hour",
    y="Revenue",
    ax=ax
)

ax.set_title("Revenue by Hour of Day")
ax.set_xlabel("Hour (24-Hour Format)")
ax.set_ylabel("Revenue")

plt.tight_layout()
plt.savefig(
    output_dir/"Sales_By_Hour.png",
    dpi = 300,
    bbox_inches = "tight"
)
plt.show()
```

```python
## Compare Python and MySQL totals
```

```python
with engine.connect() as connection:
    sql_revenue = connection.execute(
        text("""
            SELECT ROUND(SUM(Revenue), 2)
            FROM retail_transactions
        """)
    ).scalar()

    sql_invoices = connection.execute(
        text("""
            SELECT COUNT(DISTINCT InvoiceNo)
            FROM retail_transactions
        """)
    ).scalar()

python_revenue = round(df["Revenue"].sum(), 2)
python_invoices = df["InvoiceNo"].nunique()

print("Python revenue:", python_revenue)
print("SQL revenue:", float(sql_revenue or 0))

print("Python invoices:", python_invoices)
print("SQL invoices:", sql_invoices)

assert abs(python_revenue - float(sql_revenue or 0)) < 0.01
assert python_invoices == sql_invoices

print("Validation successful!")
```
